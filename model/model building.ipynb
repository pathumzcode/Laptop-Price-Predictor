```python
def add_company(input):
    other_companies = {
        "Samsung", "Razer", "Mediacom", "Microsoft", "Xiaomi",
        "Vero", "Chuwi", "Google", "Fujitsu", "LG", "Huawei"
    }

    if input in other_companies:
        return "Other"
    return input


data["Company"] = data["Company"].apply(add_company)
```

```python
def set_processor(name):
    if name == "Intel Core i7" or name == "Intel Core i5" or name == "Intel Core i3":
        return name
    else:
        if name.split()[0] == "AMD":
            return "AMD"
        else:
            return "Other"

data['cpu_name'] = data['cpu_name'].apply(set_processor)
```

```python
def set_os(input):
    if input == "Windows 10" or input == "Windows 7" or input == "Windows 10 S":
        return "Windows"
    elif input == "macOS" or input == "Mac OS X":
        return "Mac"
    elif input == "Linux" or input == "Linux":
        return "Linux"
    else:
        return "Other"

data['OpSys'] = data['OpSys'].apply(set_os )
```

Feature Engineeering 

```python
x = data.drop('Price_euros', axis=1)
y = data['Price_euros']
```

```python
!pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
def model_acc(model):
    model.fit(x_train, y_train)
    acc = model.score(x_test, y_test)
    print(str(model) + " Accuracy: " + str(acc * 100) + "%")
```

```python
# Convert categorical columns to numeric before fitting the model
x_train = pd.get_dummies(x_train)
x_test = pd.get_dummies(x_test)

# Align columns to ensure both train and test sets have the same features
x_train, x_test = x_train.align(x_test, join="outer", axis=1, fill_value=0)

from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
model_acc(lr)

from sklearn.linear_model import Lasso
lasso = Lasso()
model_acc(lasso)

from sklearn.tree import DecisionTreeRegressor
dtr = DecisionTreeRegressor()
model_acc(dtr)

from sklearn.ensemble import RandomForestRegressor
rfr = RandomForestRegressor()
model_acc(rfr)
```

```python
from sklearn.model_selection import GridSearchCV

parameaters = {'n_estimators': [10, 50, 100],
                'criterion': ['squared_error', 'absolute_error', 'poisson']}

grid_obj = GridSearchCV(estimator=rfr, param_grid=parameaters)

grid_fit = grid_obj.fit(x_train, y_train)

best_model = grid_fit.best_estimator_
best_model
```

```python
best_model.score(x_test, y_test)
```

```python
import pickle
with open('predictor.pickle', 'wb') as file:
    pickle.dump(best_model, file)
```
